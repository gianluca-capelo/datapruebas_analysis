# Resultados ML — Mar 2026

```python
import pandas as pd

FILE_NEW = "../../analysis/results/consolidated/combined_summary_with_dispersion_2026-03-06_2028_2026-03-07_1213.csv"

df = pd.read_csv(FILE_NEW)
print(f"{len(df)} filas cargadas")
df.head(3)
```

```python
# sort_p = 0 para p_value < 0.001 → primeros dentro de cada dataset, ordenados solo por mae
# sort_p = p_value real para el resto → ordenados por p_value luego mae
df["sort_p"] = df["p_value_mae"].apply(lambda v: v if v >= 0.001 else 0)
df_sorted = (
    df.drop(columns=["y_true", "y_pred", "mse", "r2"])
    .sort_values(["dataset", "sort_p", "mae"])
    .drop(columns=["sort_p"])
    .reset_index(drop=True)
)

pd.set_option("display.max_rows", len(df_sorted))
df_sorted
```

```python
SUBSET_MAP = {
    "tmt_k_mean":   "TMT ∩ CDT",
    "tmt_k4":       "TMT ∩ CDT",
    "tmt_k6":       "TMT ∩ CDT",
    "tmt_k6_v2":    "TMT ∩ CDT",
    "tmt_dprime":   "TMT ∩ GNG",
    "tmt_accuracy": "TMT ∩ GNG",
    "tmt_fa":       "TMT ∩ GNG",
    "tmt_eficiencia": "TMT ∩ GNG",
    "tmt_c":        "TMT ∩ GNG",
    "tmt_ssrt":     "TMT ∩ SST",
    "tmt_age":      "TMT",
}

df_sorted.insert(df_sorted.columns.get_loc("dataset") + 1, "Subset", df_sorted["dataset"].map(SUBSET_MAP))
df_sorted
```

```python
df_display = df_sorted.copy()
df_display["mae"]         = df_display["mae"].round(2)
df_display["sd_y_true"]   = df_display["sd_y_true"].round(2)
df_display["iqr_y_true"]  = df_display["iqr_y_true"].round(2)
df_display["p_value_mae"] = df_display["p_value_mae"].apply(
    lambda v: "<0.001" if v < 0.001 else f"{v:.3f}"
)

df_display
```

```python
df_display = df_display.rename(columns={
    "target":       "Target",
    "Subset":       "Subset",
    "model":        "Model",
    "mae":          "MAE",
    "sd_y_true":    "SD_Target",
    "iqr_y_true":   "IQR_Target",
    "p_value_mae":  "p-value",
})

df_display
```

```python
df_display.to_csv("../../analysis/results/consolidated/results_mar_2026.csv", index=False)
print("Guardado.")
```
