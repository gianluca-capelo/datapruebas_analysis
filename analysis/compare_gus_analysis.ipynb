# Comparación: Análisis Gus vs Últimas corridas

```python
import pandas as pd
from pathlib import Path

BASE = Path('/home/gianluca/Research/datapruebas_analysis')
GUS_DIR = BASE / 'analysis 27-01 gus'

tasks = {
    'SST': ('2026-01-27_19-53-13', 'sst_analysis.csv', 'sst_analysis'),
    'CDT': ('2026-01-27_19-53-16', 'cdt_analysis.csv', 'cdt_analysis'),
    'Go/No-Go': ('2026-01-27_19-53-18', 'gonogo_analysis.csv', 'gonogo_analysis')
}
```

```python
# Cargar todos los dataframes
dfs = {}
for name, (gus_ts, filename, data_dir) in tasks.items():
    latest_dir = sorted((BASE / 'data' / data_dir).iterdir())[-1]
    dfs[name] = {
        'gus': pd.read_csv(GUS_DIR / gus_ts / filename).sort_values('subject_id').reset_index(drop=True),
        'latest': pd.read_csv(latest_dir / filename).sort_values('subject_id').reset_index(drop=True),
        'latest_dir': latest_dir.name
    }
    print(f"{name}: Gus={len(dfs[name]['gus'])} | Latest={len(dfs[name]['latest'])} ({latest_dir.name})")
```

```python
# Comparar (deben ser idénticos excepto por el orden)
for name, data in dfs.items():
    gus, latest = data['gus'], data['latest']
    diff = gus.compare(latest)
    
    print(f"=== {name} ===")
    if diff.empty:
        print("Idénticos ✓\n")
    else:
        print("Diferencias encontradas:")
        display(diff)
```

## TMT

```python
# Cargar TMT
tmt_gus = pd.read_csv(GUS_DIR / '2026-01-27_17-00-02' / 'analysis.csv').sort_values(['subject_id', 'trial_order_of_appearance']).reset_index(drop=True)
latest_tmt_dir = sorted((BASE / 'data' / 'hand_analysis').iterdir())[-1]
tmt_latest = pd.read_csv(latest_tmt_dir / 'analysis.csv').sort_values(['subject_id', 'trial_order_of_appearance']).reset_index(drop=True)

print(f"TMT: Gus={len(tmt_gus)} | Latest={len(tmt_latest)} ({latest_tmt_dir.name})")
```

```python
# Ordenar columnas y comparar
tmt_latest_reord = tmt_latest[tmt_gus.columns]
diff = tmt_gus.compare(tmt_latest_reord)

if diff.empty:
    print("TMT: Idénticos ✓")
else:
    print(f"TMT: {len(diff)} filas con diferencias")
    display(diff)
```
